```python
#part 1 - Extract Unique Item Names
unique_items = set(sale["item"] for sale in sales)
unique_items
```

```python
#part 2 - Compute Total Net Revenue per Item
item_net_revenue =  {item: sum((sale["price"] * sale["quantity"]) * (1 - sale["discount"]) for sale in sales  if sale["item"] == item) for item in unique_items}
item_net_revenue
```

```python
#Part 3 — Identify High-Performing Items
top_items = [key for key,value in item_net_revenue.items() if value > 2000]
top_itemstop_items = [key for key,value in item_net_revenue.items() if value > 2000]
top_items
```

```python
#Part 4 — Filter Sales for High-Performing Items
filtered_sales = [sale for sale in sales if sale["item"] in top_items]
filtered_sales
```

```python
#Part 5 — Compute Total Quantity Sold per Branch
branch_quantities = {b: sum(s["quantity"] for s in filtered_sales if s["branch"] == b) for b in {s["branch"] for s in filtered_sales}}
branch_quantities
```

```python
#Part 6 — Determine the Most Successful Branch
most_successful_branch = max(branch_quantities, key=branch_quantities.get)
most_successful_branch
```
