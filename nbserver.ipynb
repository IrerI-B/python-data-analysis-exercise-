```python
#Part 2
item_net_revenue = { item: sum((sale["price"] * sale["quantity"]) * (1 - sale["discount"]) for sale in sales if sale["item"] == item)
 for item in unique_items
}
item_net_revenue
```

```python
#Part 3
top_items = [key for key,value in item_net_revenue.items() if value > 2000]
top_itemstop_items = [key for key,value in item_net_revenue.items() if value > 2000]
top_items
```

```python
 #part 4
filtered_sales = [sale for sale in sales if sale["item"] in top_items]
filtered_sales
```

```python
#part 5 Using filtered_sales from Part 4, compute the total quantity sold per branch
branch_quantities = {}
for sale in filtered_sales:
    branch = sale["branch"]
    quantity = sale["quantity"]
    branch_quantities [branch] = branch_quantities.get(branch, 0) + quantity

branch_quantities
```

```python
most_successful_branch = max(branch_quantities, key=branch_quantities.get)
most_successful_branch
```
